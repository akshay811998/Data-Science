# 8)Test the similarity between the degree_t(Sci&Tech) and specialisation(Mkt&HR) with respect to salary at significance level of 5%.(Make decision using Hypothesis Testing)
##### t-statistic : 2.692
##### p-value     : 0.007898
##### Reject Ho as There's a Significant difference exists between Sci&Tech and Mkt&HR salaries

```python
from scipy.stats import ttest_ind
SciTech = preprocessed[preprocessed['degree_t']=='Sci&Tech']['salary']
MktHR = preprocessed[preprocessed['specialisation']=='Mkt&HR']['salary']
ttest_ind(SciTech, MktHR)
result = ttest_ind(SciTech, MktHR)
print(result)
```

```python
# H0 : Salary of Sci&Tech and Mkt&HR are similar (no significant difference)
# H1 : Salary of Sci&Tech and Mkt&HR are NOT similar
# threshold  = 0.05

threshold = 0.05
print("t-statistic :", round(result.statistic, 4))
print("p-value     :", round(result.pvalue, 6))

if result.pvalue < threshold:
    print("Reject Ho as There's a Significant difference exists between Sci&Tech and Mkt&HR salaries")
else:
    print("Accept Ho as There's no significant difference between Sci&Tech and Mkt&HR salaries")
```

# 10)What is the probability Density Function of the salary range from 700000 to 900000?
##### Using the normal distribution fitted to the salary column (Mean=198702.326, Standard Deviation=154780.927), the probability of salary falling between ₹7,00,000 and ₹9,00,000 was computed as the area under the PDF curve using CDF(9,00,000) − CDF(7,00,000) is 0.03840460 approx.

```python
sns.distplot(preprocessed["salary"])
```

```python
def get_pdf_probability(dataset,startrange,endrange):
    from matplotlib import pyplot
    from scipy.stats import norm
    import seaborn as sns
    ax = sns.distplot(dataset,kde=True,kde_kws={'color':'blue'},color='Green')
    pyplot.axvline(startrange,color='Red')
    pyplot.axvline(endrange,color='Red')
    # generate a sample
    sample = dataset
    # calculate parameters
    sample_mean =sample.mean()
    sample_std = sample.std()
    print('Mean=%.3f, Standard Deviation=%.3f' % (sample_mean, sample_std))
    # define the distribution
    dist = norm(sample_mean, sample_std)
    
    # sample probabilities for a range of outcomes
    values = [value for value in range(startrange, endrange)]
    probabilities = [dist.pdf(value) for value in values]    
    prob=sum(probabilities)
    print("The area between range({},{}):{}".format(startrange,endrange,sum(probabilities)))
    return prob
```

```python
get_pdf_probability(dataset["salary"],70000,90000)
```

# 11)Test the similarity between the degree_t(Sci&Tech)with respect to etest_p and mba_p at significance level of 5%.(Make decision using Hypothesis Testing)
##### t-statistic : 5.005
##### p-value     : 6e-06
##### Reject Ho as There's a Significant difference exists between Sci&Tech and Mkt&HR salaries

```python
from scipy.stats import ttest_rel
ETEST = preprocessed[preprocessed['degree_t']=='Sci&Tech']['etest_p']
MBA = preprocessed[preprocessed['degree_t']=='Sci&Tech']['mba_p']
ttest_rel(ETEST, MBA)
result = ttest_rel(ETEST, MBA)
print(result)
```

```python
# H0 : Salary of Sci&Tech and Mkt&HR are similar (no significant difference)
# H1 : Salary of Sci&Tech and Mkt&HR are NOT similar
# threshold  = 0.05

threshold = 0.05
print("t-statistic :", round(result.statistic, 4))
print("p-value     :", round(result.pvalue, 6))

if result.pvalue < threshold:
    print("Reject Ho as There's a Significant difference exists between Sci&Tech and Mkt&HR salaries")
else:
    print("Accept Ho as There's no significant difference between Sci&Tech and Mkt&HR salaries")
```
